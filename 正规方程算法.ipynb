```python
import numpy as np
```

```python
class LinearRegression:
    def __init__(self):
        self.w = None

    def fit(self, X, y):
        m = X.shape[0]
        Xb = np.hstack([np.ones((m, 1)), X])
        self.w = np.linalg.inv(Xb.T @ Xb) @ Xb.T @ y
        return self

    def predict(self, X):
        m = X.shape[0]
        Xb = np.hstack([np.ones((m, 1)), X])
        return Xb @ self.w
```

```python

np.random.seed(42)
m = 100
X = 2 * np.random.rand(m, 1)
y = 4 + 3 * X[:, 0] + np.random.randn(m) * 0.5
model = LinearRegression().fit(X, y)
print(f"学到的参数 w（真实值 [4, 3]）: {model.w}")
```
